# Mission 7：建立模擬的大學生學習資料

```python
import numpy as np
import pandas as pd

student_count = 60      # Mission 2 的學生人數（把 # 拿掉才有效）
random_seed   = 42      # ★ 請改成你 Mission 6 實際使用的那個種子，才能重現相同資料

# ---------- 以下原本內容不變 ----------
n = student_count
rng = np.random.default_rng(random_seed)

student_id = np.arange(1, n + 1)

departments = ["資訊管理", "企業管理", "金融管理"]
department = rng.choice(departments, size=n)

sleep_hours_raw = rng.normal(loc=6.7, scale=1.1, size=n)
sleep_hours = np.round(np.clip(sleep_hours_raw, 0, 24), 1)

screen_time_hours_raw = rng.normal(loc=5.0, scale=1.5, size=n)
screen_time_hours = np.round(np.clip(screen_time_hours_raw, 1.0, 10.0), 1)

study_hours_raw = rng.normal(loc=2.5, scale=1.0, size=n)
study_hours = np.round(np.clip(study_hours_raw, 0.5, 6.0), 1)

attendance_rate_raw = rng.normal(loc=88.0, scale=8.0, size=n)
attendance_rate = np.round(np.clip(attendance_rate_raw, 60, 100)).astype(int)

df = pd.DataFrame({
    "student_id": student_id,
    "department": department,
    "sleep_hours": sleep_hours,
    "screen_time_hours": screen_time_hours,
    "study_hours": study_hours,
    "attendance_rate": attendance_rate
})

print("前五筆學習資料：")
df.head()
```

# Mission 8：建小考成績欄位

```python
import numpy as np
import pandas as pd

n = 100
rng = np.random.default_rng(42)
df = pd.DataFrame({
    "student_id":list(range(1, n + 1)),
    "study_hours": rng.uniform(0, 10, n).round(1),
    "sleep_hours": rng.uniform(4, 10, n).round(1),
    "screen_time_hours": rng.uniform(0, 8, n).round(1),
    "attendance_rate": rng.uniform(50, 100, n).round(0),
})

# 步驟 1：基礎分數
base = 50
# 步驟 2：各項影響
study_effect      = 3 * df["study_hours"]
sleep_effect      = 2 * df["sleep_hours"]
screen_effect     = -1.2 * df["screen_time_hours"]
attendance_effect = 0.15 * df["attendance_rate"]
# 步驟 3：隨機誤差
error = rng.normal(loc=0.0, scale=5.0, size=n)
# 步驟 4：加總
raw_score = base + study_effect + sleep_effect + screen_effect + attendance_effect + error
# 步驟 5：限制 0~100 並轉整數
quiz_score = np.clip(raw_score, 0, 100)
quiz_score = np.round(quiz_score).astype(int)
# 步驟 6：寫回 DataFrame
df["quiz_score"] = quiz_score

display_cols = ["student_id", "study_hours", "sleep_hours",
                "screen_time_hours", "attendance_rate", "quiz_score"]
print("關鍵欄位前五筆：")
df[display_cols].head()
```
