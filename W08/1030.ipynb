# Mission 4：算術運算與成績公式

```python
# 假想學生的基本資料
study_hours = 2.5      # 讀書時數（小時）
sleep_hours = 7.0      # 睡眠時數（小時）
screen_time_hours = 4.0 # 螢幕使用時數（小時）
attendance_rate = 90   # 出席率（%）

# 分別計算各項對成績的影響
base_score = 50
study_effect = 3 * study_hours
sleep_effect = 2 * sleep_hours
screen_effect = -1.2 * screen_time_hours   # 負數表示扣分
attendance_effect = 0.15 * attendance_rate

# 合計成績（尚未四捨五入）
raw_score = base_score + study_effect + sleep_effect + screen_effect + attendance_effect
print("未四捨五入的原始成績:", raw_score)

# 四捨五入成整數（0~100 之間）
quiz_score = int(round(raw_score))
# 確保不超過 0~100
if quiz_score < 0:
    quiz_score = 0
if quiz_score > 100:
    quiz_score = 100
print("最終小考成績:", quiz_score)
```

# Mission 5：條件判斷與自訂函式

```python
def describe_score(score):
    """根據分數回傳評語"""
    if score >= 80:
        return "表現良好"
    elif score >= 60:
        return "達到基本標準"
    else:
        return "需要加強"

# 測試不同分數
test_scores = [92, 75, 58, 40]
for s in test_scores:
    result = describe_score(s)
    print(f"分數 {s} -> {result}")
```

# Mission 7：建立模擬的大學生學習資料

```python
import numpy as np
import pandas as pd

student_count = 60      # Mission 2 的學生人數（把 # 拿掉才有效）
random_seed   = 42      # ★ 請改成你 Mission 6 實際使用的那個種子，才能重現相同資料

# ---------- 以下原本內容不變 ----------
n = student_count
rng = np.random.default_rng(random_seed)

student_id = np.arange(1, n + 1)

departments = ["資訊管理", "企業管理", "金融管理"]
department = rng.choice(departments, size=n)

sleep_hours_raw = rng.normal(loc=6.7, scale=1.1, size=n)
sleep_hours = np.round(np.clip(sleep_hours_raw, 0, 24), 1)

screen_time_hours_raw = rng.normal(loc=5.0, scale=1.5, size=n)
screen_time_hours = np.round(np.clip(screen_time_hours_raw, 1.0, 10.0), 1)

study_hours_raw = rng.normal(loc=2.5, scale=1.0, size=n)
study_hours = np.round(np.clip(study_hours_raw, 0.5, 6.0), 1)

attendance_rate_raw = rng.normal(loc=88.0, scale=8.0, size=n)
attendance_rate = np.round(np.clip(attendance_rate_raw, 60, 100)).astype(int)

df = pd.DataFrame({
    "student_id": student_id,
    "department": department,
    "sleep_hours": sleep_hours,
    "screen_time_hours": screen_time_hours,
    "study_hours": study_hours,
    "attendance_rate": attendance_rate
})

print("前五筆學習資料：")
df.head()
```

# Mission 8：建小考成績欄位

```python
import numpy as np
import pandas as pd

n = 100
rng = np.random.default_rng(42)
df = pd.DataFrame({
    "student_id":list(range(1, n + 1)),
    "study_hours": rng.uniform(0, 10, n).round(1),
    "sleep_hours": rng.uniform(4, 10, n).round(1),
    "screen_time_hours": rng.uniform(0, 8, n).round(1),
    "attendance_rate": rng.uniform(50, 100, n).round(0),
})

# 步驟 1：基礎分數
base = 50
# 步驟 2：各項影響
study_effect      = 3 * df["study_hours"]
sleep_effect      = 2 * df["sleep_hours"]
screen_effect     = -1.2 * df["screen_time_hours"]
attendance_effect = 0.15 * df["attendance_rate"]
# 步驟 3：隨機誤差
error = rng.normal(loc=0.0, scale=5.0, size=n)
# 步驟 4：加總
raw_score = base + study_effect + sleep_effect + screen_effect + attendance_effect + error
# 步驟 5：限制 0~100 並轉整數
quiz_score = np.clip(raw_score, 0, 100)
quiz_score = np.round(quiz_score).astype(int)
# 步驟 6：寫回 DataFrame
df["quiz_score"] = quiz_score

display_cols = ["student_id", "study_hours", "sleep_hours",
                "screen_time_hours", "attendance_rate", "quiz_score"]
print("關鍵欄位前五筆：")
df[display_cols].head()
```

# Mission 9：加入缺失值與異常值

```python
import numpy as np
import pandas as pd

# 建立或讀取資料框 df
np.random.seed(42)  # 固定隨機種子，使每次執行結果可重現
n_rows = 20  # 至少要有 18 筆（索引 0~17），這裡多給幾筆以免越界
df = pd.DataFrame({
    "student_id": [f"S{i:03d}" for i in range(1, n_rows + 1)],   # S001 ~ S020
    "sleep_hours": np.random.uniform(5, 9, size=n_rows),        # 睡眠時數 5~9 小時
    "screen_time_hours": np.random.uniform(2, 6, size=n_rows),  # 螢幕時數 2~6 小時
    # 你可以在此加入其他欄位，例如：gender, age, gpa 等等
})

df.loc[[5, 17], "sleep_hours"] = np.nan
df.loc[8, "screen_time_hours"] = 15.0

display(df.loc[[5, 8, 17], ["student_id", "sleep_hours", "screen_time_hours"]])
```
